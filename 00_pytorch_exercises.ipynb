```python
torch.manual_seed(1234)
rand_tensor_A = torch.rand(2, 3)
torch.manual_seed(1234)
rand_tensor_B = torch.rand(2, 3)

print(rand_tensor_A)
print(rand_tensor_B)
```

```python
tensor = torch.matmul(rand_tensor_A, rand_tensor_B.T)
tensor
```

```python
print(torch.min(tensor), torch.max(tensor))
```

```python
print(tensor.argmax())
print(tensor.argmin())
```
